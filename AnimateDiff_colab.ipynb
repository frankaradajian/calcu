```python
# Quick test render using Stable Diffusion pipeline
from diffusers import StableDiffusionPipeline
import torch

pipe = StableDiffusionPipeline.from_pretrained("runwayml/stable-diffusion-v1-5").to("cuda")

prompt = "Camel caravan trekking through desert dunes, cinematic motion"
image = pipe(prompt, height=512, width=512).images[0]

image.save("/content/caravan_frame.png")
```

```python
import imageio

frames = []
for i in range(240):  # ~10 seconds at 24fps
    img = pipe(prompt, height=512, width=512).images[0]
    frames.append(img)

# Save as MP4
output_path = "/content/caravan.mp4"
imageio.mimsave(output_path, frames, fps=24)

print(f"Video saved to {output_path}")
```

```python
import imageio
import os

prompt = "Camel caravan trekking through desert dunes, cinematic motion"
output_dir = "/content/caravan_frames"
os.makedirs(output_dir, exist_ok=True)

frames = []
for i in range(240):  # total frames
    img = pipe(prompt, height=512, width=512).images[0]
    frame_path = f"{output_dir}/frame_{i:03d}.png"
    img.save(frame_path)   # save each frame individually
    frames.append(img)

    # Checkpoint every 50 frames
    if (i+1) % 50 == 0:
        partial_path = f"/content/caravan_partial_{i+1}.mp4"
        imageio.mimsave(partial_path, frames, fps=24)
        print(f"Checkpoint saved: {partial_path}")
```

```python
import imageio
import glob

frame_files = sorted(glob.glob("/content/caravan_frames/frame_*.png"))
frames = [imageio.imread(f) for f in frame_files]

imageio.mimsave("/content/caravan_resume.mp4", frames, fps=24)
print("Resumed video saved to /content/caravan_resume.mp4")
```
